# 01 MIROC6

```python
model_import = 'MIROC6'
model_export = 'miroc6'
```

#### 01.01 Historical Period

```python
#### Load data
print(f"Processing model: {model_export} - Historical Period")
os.makedirs(f'outputs/cmip6_models/{model_import}/historical', exist_ok=True)

path_hist_gcm_model = glob.glob(f'/lustre/geocean/WORK/users/alonsoap/personal/mini_projects/stochastic_gcms_nc/data_espg/tos_Oday_{model_import}_historical_r1i1p1f1_gn_*.nc')

ds_hist_gcm_model = xr.open_mfdataset(path_hist_gcm_model, chunks={"time": 365})[["tos"]].rename({"tos": "sst"})#.rename({"lon": "longitude", "lat": "latitude", "tos": "sst"})
ds_hist_gcm_model["time"] = ds_hist_gcm_model.time.astype("datetime64[ns]")
ds_hist_gcm_model = ds_hist_gcm_model.resample(time="1D").mean()

ds = ds_hist_gcm_model
lat_1d = ds.latitude[:, 0].data
lon_1d = ds.longitude[0, :].data
ds = ds.assign_coords(latitude=("y", lat_1d),longitude=("x", lon_1d))
ds = ds.swap_dims({"y": "latitude", "x": "longitude"})
ds = ds.sortby("latitude")
ds = ds.sortby("longitude")
ds_hist_gcm_model_regrid = ds.sel(latitude=slice(10, 30),longitude=slice(260, 350))

ds_hist_gcm_model_regrid.to_netcdf(f'outputs/cmip6_models/{model_export}/historical/sst_{model_export}_historical_regrid.nc', mode='w')
```

# 02. EC-Earth3-Veg_LR

```python
model_import = 'EC-Earth3-Veg'
model_export = 'ec_earth3_veg_lr'
```

#### 02.01 Historical Period

```python
#### Load data
print(f"Processing model: {model_export} - Historical Period")
os.makedirs(f'outputs/cmip6_models/{model_export}/historical', exist_ok=True)

path_hist_gcm_model = glob.glob(f'/lustre/geocean/WORK/users/alonsoap/personal/mini_projects/stochastic_gcms_nc/data_espg/tos_Oday_{model_import}_historical_r1i1p1f1_gn_*.nc')

ds_hist_gcm_model = xr.open_mfdataset(path_hist_gcm_model, chunks={"time": 365})[["tos"]].rename({"tos": "sst"})#.rename({"lon": "longitude", "lat": "latitude", "tos": "sst"})

ds = ds_hist_gcm_model
ds = ds.rename({"latitude": "lat","longitude": "lon"})
ds_out = xe.util.grid_2d(lon0_b=260, lon1_b=350, d_lon=1.0,lat0_b=10,  lat1_b=30,  d_lat=1.0)
regridder = xe.Regridder(ds, ds_out, "bilinear", ignore_degenerate=True)
ds_regrid = regridder(ds)
ds_regrid = ds_regrid.rename({'lat': 'latitude', 'lon': 'longitude'})
ds = ds_regrid

lat_1d = ds.latitude[:, 0].data
lon_1d = ds.longitude[0, :].data
ds_regrid = ds_regrid.assign_coords(latitude=("y", lat_1d),longitude=("x", lon_1d))
ds_regrid = ds_regrid.swap_dims({"y": "latitude", "x": "longitude"})
ds_regrid = ds_regrid.sortby("latitude")
ds_regrid = ds_regrid.sortby("longitude")

ds_regrid.to_netcdf(f'outputs/cmip6_models/{model_export}/historical/sst_{model_export}_historical_regrid.nc', mode='w')
```
